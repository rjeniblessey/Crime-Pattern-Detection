Handling Missing Values in Numerical Columns

```python
import numpy as np 

# Step 1: Fill median for numerical columns EXCEPT latitude/longitude
numeric_columns = data.select_dtypes(include=np.number).columns

numeric_columns = numeric_columns.drop(
    ["Latitude", "Longitude"],
    errors="ignore"
)

for col in numeric_columns:
    data[col] = data[col].fillna(data[col].median())

```

```python
data.isnull().sum()
```

Drop Null Values in Latitude and Longitude

```python

# Step 2: Remove records with missing coordinates
data = data.dropna(
    subset=["Latitude", "Longitude"]
)
```

```python
print(data[["Latitude", "Longitude"]].isnull().sum())
```

Handling Missing Values in Categorical Columns

```python
categorical_columns = data.select_dtypes(
    include="object"
).columns

for col in categorical_columns:
    data[col] = data[col].fillna("Unknown")
```

```python
data.isnull().sum()
```

```python
# Latitude  → 41–43
# Longitude → -88–-87

if "Latitude" in data.columns and "Longitude" in data.columns:

    df = data[
        (data["Latitude"].between(41, 43)) &
        (data["Longitude"].between(-88, -87))
    ]
```

Encode Crime Type

```python
# One-hot Encoding

crime_encoded = pd.get_dummies(
    data["Primary Type"],
    prefix="Crime"
)
crime_encoded.head()
```

Convert Boolean Values to Integers

```python
crime_encoded = crime_encoded.astype(int)

# Combine encoded columns
data = pd.concat(
    [data, crime_encoded],
    axis=1
)
```

Select Clustering Features

```python
crime_columns = [
    col for col in data.columns
    if col.startswith("Crime_")
]
```

```python
features = [
    "Latitude",
    "Longitude",
    "Hour",
    "DayOfWeek",
    "Month"
] + crime_columns

X = data[features].copy()
print(X.shape)
print()
print(X.head())
```

Standardize the Features

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[:5])
```

Find the Optimal Number of Clusters

```python
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
```

Using Elbow Method

```python
inertia = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    inertia.append(
        kmeans.inertia_
    )
```

Visualize

```python
plt.figure(figsize=(10, 6))

plt.plot(
    range(2, 11),
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.show()
```

Silhouette Score

```python
sample = X_scaled[
    np.random.choice(
        X_scaled.shape[0],
        size=min(10000, X_scaled.shape[0]),
        replace=False
    )
]
silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(sample)

    score = silhouette_score(
        sample,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"K = {k}, "
        f"Silhouette Score = {score:.4f}"
    )
```

Visualize

```python
plt.figure(figsize=(10, 6))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score")

plt.show()
```

Apply K-Means

```python
optimal_k = 4
kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

# Fit the model
data["Cluster"] = kmeans.fit_predict(
    X_scaled
)
```

Count crimes in each cluster

```python
cluster_counts = (
    data["Cluster"]
    .value_counts()
    .sort_index()
)

print(cluster_counts)
```

Visualize

```python
plt.figure(figsize=(8, 5))

cluster_counts.plot(
    kind="bar"
)

plt.xlabel("Cluster")
plt.ylabel("Number of Crimes")
plt.title("Crime Count by Cluster")

plt.show()
```

Analyze Each Cluster

```python
for cluster in sorted(
    data["Cluster"].unique()
):

    cluster_data = data[
        data["Cluster"] == cluster
    ]

    print("\n======================")
    print("Cluster:", cluster)
    print("======================")

    print(
        "Number of crimes:",
        len(cluster_data)
    )

    print("\nTop crime types:")

    print(
        cluster_data[
            "Primary Type"
        ]
        .value_counts()
        .head(5)
    )

    print(
        "\nAverage latitude:",
        cluster_data["Latitude"].mean()
    )

    print(
        "Average longitude:",
        cluster_data["Longitude"].mean()
    )

    print(
            "Average crime hour:",
            cluster_data["Hour"].mean()
        )
    
```

Find the dominant crime in each cluster

```python
dominant_crime = (
    data.groupby("Cluster")[
        "Primary Type"
    ]
    .agg(
        lambda x:
        x.value_counts().index[0]
    )
)
print(dominant_crime)
```

Final Summary Table
